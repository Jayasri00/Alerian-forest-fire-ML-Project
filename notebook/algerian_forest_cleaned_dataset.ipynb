```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
pd.set_option('display.max_rows',None)
dataset=pd.read_csv("Algerian_forest_fires_dataset.csv")
dataset
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
## missing values
dataset.isnull().sum()
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
##the dataset is converted into 2 sets based on region from 122th index,we can make 
##a column based on the region
##1.bejaia ragion dataset
##2.sidi-bel abbes region dataset
##add new column with region
```

```python
dataset.loc[:122,'Region']=0
dataset.loc[122:,'Region']=1
df=dataset
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## remove null values
df=df.dropna().reset_index(drop=True)
df.head()
```

```python
df.isnull().sum()
```

```python
 df.iloc[[122]]
```

```python
##removeing 122nd row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## we have to remove the space in classes column
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

#### changes the required columns as integer data type

```python
df[['day','month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day','month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

##changes the other columns  to float  data type

```python
objects=[features for features in df.columns if df[features].dtype=='O']
objects
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
 ## let save the cleaned dataset
df.to_csv("algerian_forest_cleaned_dataset_csv")
```

##monthly fire analysis


```python
df['Classes']=np.where(df['Classes'].str.contains('not fire'),'not fire','fire');
df.head()
```

```python
dftemp=df.loc[df['Region']==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.ylabel("number of fires",weight='bold')
plt.xlabel("months",weight='bold')
plt.title("fire analysis of bejaia regions",weight='bold') 
```

```python
dftemp=df.loc[df['Region']==0]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.ylabel("number of fires",weight='bold')
plt.xlabel("months",weight='bold')
plt.title("fire analysis of sidi-bel regions",weight='bold') 
```
